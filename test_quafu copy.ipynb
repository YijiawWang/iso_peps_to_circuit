```python
# circuit_name = 'gates_2patterns' # gates_2patterns
# with open(f'/data/miyan/isopeps_deco/iso_peps_to_circuit/gates_2patterns/final_circuit/circuit.qasm', 'r') as f:
#     circuit = f.read()
# print(circuit)

circuit = '''
OPENQASM 2.0;
include "qelib1.inc";

qreg q[9];
creg c[9];

// 1. 在第 0 个比特上施加 Hadamard 门，创造叠加态
h q[0];

// 2. 依次施加 CNOT 门，将纠缠态像链条一样传递
cx q[0],q[1];
cx q[1],q[2];
cx q[2],q[3];
cx q[3],q[4];
cx q[4],q[5];
cx q[5],q[6];
cx q[6],q[7];
cx q[7],q[8];

// 3. 测量所有比特
measure q[0] -> c[0];
measure q[1] -> c[1];
measure q[2] -> c[2];
measure q[3] -> c[3];
measure q[4] -> c[4];
measure q[5] -> c[5];
measure q[6] -> c[6];
measure q[7] -> c[7];
measure q[8] -> c[8];
'''
```

```python
task = {
  'chip': 'Baihua',  # chip name
  'name': '2-patterns',  # task name
  'circuit': circuit, # circuit written in OpenQASM2.0
  'shots': 4096, # an integer multiple of 1024
  'options':{
    'compiler': 'qiskit', # defaults to 'quarkcircuit'
    'correct': False, # readout error correction 
    'open_dd': None, # dynamial decoupling,  # defaults to None
    # 'target_qubits': [25, 31, 38, 30, 37, 43, 36, 42, 49] # [0, 1]
  }
}
```

```python
tid = tmgr.run(task)
print(tid) # tid means task id
# output: 2024041917095371986
```
